# Harvest viewer
__See what the harvester did, grain by grain__

Click a grain on the map, then a reflection (row) of its sinogram, then a scan in its I(dty) plot. The reflection is integrated again from the raw frames with the harvester's own kernel and its boxes are shown frame by frame, with the other grains' boxes. The `results` menu switches between the sets of results given in `sources`: a folder written by `harvest_1_run.ipynb`, or a single harvest file.
